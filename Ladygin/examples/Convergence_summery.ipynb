```python
import numpy as np
import matplotlib.pyplot as plt
```

### 1. Al (1 atom) metal

```python
mat = "Al"

data = np.genfromtxt(f'./calcs/{mat}/stats.dat', skip_header = 1)

plt.figure(figsize = (20, 10))


ax1 = plt.subplot(121)


ax1.plot(data[:, 0], data[:, 1])

ax1.set_ylabel("Energy (eV)", fontsize = 15)
ax1.set_xlabel("kpoint side (k x k x k)", fontsize = 15)

ax1.set_xlim(0, 100)

ax2 = plt.subplot(122)


ax2.plot(data[:, 0], np.log10(data[:, 2]*1e3))
ax2.plot(range(101), [-2]*101, ls = '--')

ax2.set_ylabel("$\log_{10}$ (Error)", fontsize = 15)
ax2.set_xlabel("kpoint side (k x k x k)", fontsize = 15)

ax2.set_xlim(0, 100)
```

```python
print(f"Optimal value for Al of kpoint grid is {int(data[-1, 0])}x{int(data[-1, 0])}x{int(data[-1, 0])} with error of {data[-1, -1]*1e3} meV")
```
